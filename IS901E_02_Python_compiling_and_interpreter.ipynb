### Executing commands in the conda environment without activation

Instead of `!conda activate py36`, we can run packages directly using the path to the environment's python or by using `conda run -n py36`.

```python
# Check the python version inside the py36 environment
!/usr/local/envs/py36/bin/python --version
```

```python
# Use pip inside py36 to install uncompyle6 for decompiling older formats
!/usr/local/envs/py36/bin/pip install uncompyle6
```

```python
# Run uncompyle6 from the Python 3.6 conda environment
!/usr/local/envs/py36/bin/uncompyle6 hello.cpython-313.pyc
```
